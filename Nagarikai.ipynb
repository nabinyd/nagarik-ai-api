### PROCESSED_DIR

Purpose: This is where you store the cleaned and converted text from each PDF.

What it contains: Each PDF is converted to JSON with cleaned text, usually one JSON per PDF.

### CHUNK_DIR (chunks_legal)

Purpose: This is where the text is split into smaller chunks for easier processing in AI/ML or NLP workflows.

Example:
If a PDF has 5000 words, it can be split into chunks of 500–1000 words each:

```
chunks_legal/land-management_file1_chunk1.json
chunks_legal/land-management_file1_chunk2.json
chunks_legal/land-management_file2_chunk1.json
```

Why it exists:
Some AI models, embeddings, or search engines cannot handle very long text at once. Breaking text into chunks allows:

Efficient processing

Better search results

Easier indexing in a vector database

````python
# 2️⃣ Imports
import os
import fitz
from tqdm import tqdm
import json
import re
from pdfminer.high_level import extract_text
from preeti_unicode import convert_text

# 3️⃣ Directory Setup
BASE_DIR = "/content/drive/MyDrive/nagrik-ai-data"

PROCESSED_DIR = os.path.join(BASE_DIR, "processed_pdfs")
CHUNK_DIR = os.path.join(BASE_DIR, "chunks_legal")

os.makedirs(PROCESSED_DIR, exist_ok=True)
os.makedirs(CHUNK_DIR, exist_ok=True)

print("Processed dir:", PROCESSED_DIR)
print("Chunk dir:", CHUNK_DIR)
````

````python
def clean_text(text):
  text = convert_text(text=text, convert_numbers=True)
  text = re.sub(r"\n+", "\n", text)
  text = re.sub(r"\s+", " ", text)

  return text.strip()

def chunk_text(text, chunk_size=1000):
  words = text.split()
  chunks = []

  for i in range(0, len(words), chunk_size):
    chunk = " ".join(words[i:i+chunk_size])
    chunks.append(chunk)

  return chunks
````

````python
import re

def nepali_legal_chunker(text):
    """
    Split Nepali legal documents into meaningful sections
    based on common legal headings.
    """

    # Detect legal section headers
    pattern = r"(भाग\s*[०-९\d]+|परिच्छेद\s*[०-९\d]+|धारा\s*[०-९\d]+|उपधारा\s*[०-९\d]+|खण्ड\s*\(?[क-ह]\)?|अनुसूची\s*[०-९\d]+|\n\d+\.|\n[०-९]+\.)"

    parts = re.split(pattern, text)

    chunks = []
    current_chunk = ""

    for part in parts:

        if re.match(pattern, part.strip()):

            if current_chunk.strip():
                chunks.append(current_chunk.strip())

            current_chunk = part.strip()

        else:
            current_chunk += " " + part.strip()

    if current_chunk.strip():
        chunks.append(current_chunk.strip())

    return chunks
````

````python
def extract_legal_metadata(text):
  metadata = {}

  act_match = re.search(r"([\u0900-\u097F\s]+ऐन),?\s*(20\d{2}|२०\d{2})", text)
  if act_match:
    metadata["act_name"] = act_match.group(1).strip()
    metadata["act_year"] = act_match.group(2)

  year_match = re.search(r"(20\d{2}|२०\d{2})", text)
  if year_match:
    metadata["year"] = year_match.group()

  # Part (भाग)
  part_match = re.search(r"भाग\s*[०-९\d]+", text)
  if part_match:
    metadata["part"] = part_match.group()

  chapter_match = re.search(r"परिच्छेद\s*[०-९\d]+", text)
  if chapter_match:
    metadata["chapter"] = chapter_match.group()

  # Section (धारा)
  section_match = re.search(r"धारा\s*[०-९\d]+", text)
  if section_match:
    metadata["section"] = section_match.group()

  # Subsection (उपधारा)
  subsection_match = re.search(r"उपधारा\s*[०-९\d]+", text)
  if subsection_match:
    metadata["subsection"] = subsection_match.group()

  # Clause (खण्ड)
  clause_match = re.search(r"खण्ड\s*\(?[क-ह]\)?", text)
  if clause_match:
    metadata["clause"] = clause_match.group()

  # Schedule (अनुसूची)
  schedule_match = re.search(r"अनुसूची\s*[०-९\d]+", text)
  if schedule_match:
    metadata["schedule"] = schedule_match.group()

  return metadata
````

````python
def chunk_all_files():
    for filename in os.listdir(PROCESSED_DIR):
        if filename.endswith(".json"):
            file_path = os.path.join(PROCESSED_DIR, filename)

            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            text = data["text"]

            chunks = nepali_legal_chunker(text)

            structured_chunks = []

            for i, chunk in enumerate(chunks):
                metadata = extract_legal_metadata(chunk)

                structured_chunks.append({
                    "chunk_id": f"{filename}_{i}",
                    "source": filename,
                    "metadata": metadata,
                    "content": chunk
                })

            output_path = os.path.join(CHUNK_DIR, filename)

            with open(output_path, "w", encoding="utf-8") as f:
                json.dump(structured_chunks, f, ensure_ascii=False, indent=2)

            print(f"Chunked {filename}")

    print("Legal chunking complete.")


chunk_all_files()
````

````python
import os
import json

BASE_DIR = "/content/drive/MyDrive/nagrik-ai-data"
CHUNK_DIR = os.path.join(BASE_DIR, "chunks_legal")

chunk_texts = []
chunk_objects = []

for filename in os.listdir(CHUNK_DIR):
    if not filename.endswith(".json"):
        continue

    file_path = os.path.join(CHUNK_DIR, filename)

    try:
        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

            # Case 1: file contains a list of chunks
            if isinstance(data, list):
                for chunk in data:
                    # Safely get content
                    content = chunk.get("content") or chunk.get("text")
                    if content:
                        chunk_objects.append(chunk)
                        chunk_texts.append(content)

            # Case 2: file contains a single chunk (dict)
            elif isinstance(data, dict):
                content = data.get("content") or data.get("text")
                if content:
                    chunk_objects.append(data)
                    chunk_texts.append(content)

            else:
                print(f"Skipping {filename}: Unexpected JSON structure.")

    except Exception as e:
        print(f"Error reading {filename}: {e}")

print("Total chunks loaded:", len(chunk_objects))
````

````python
from sentence_transformers import SentenceTransformer



model = SentenceTransformer("intfloat/multilingual-e5-base")

print("Embedding model loaded.")

texts = ["passage: " + t for t in chunk_texts]

print("Texts prepared for embedding.")

embeddings = model.encode(
    texts,
    batch_size=16,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embeddings shape:", embeddings.shape)
````

````python
# ================================
# Paths
# ================================
BASE_DIR = "/content/drive/MyDrive/nagrik-ai-data"
CHUNK_DIR = os.path.join(BASE_DIR, "chunks_legal")
VECTOR_DIR = os.path.join(BASE_DIR, "vector_store")
os.makedirs(VECTOR_DIR, exist_ok=True)

FAISS_PATH = os.path.join(VECTOR_DIR, "nagarik_index.faiss")
META_PATH = os.path.join(VECTOR_DIR, "chunks_metadata.json")
INFO_PATH = os.path.join(VECTOR_DIR, "index_info.json")
````

````python
import faiss
# ================================
# Build FAISS index
# ================================
dimension = embeddings.shape[1]
index = faiss.IndexFlatIP(dimension)  # cosine similarity (normalized)
index.add(embeddings)

print("\nFAISS index built successfully")
print("Total vectors stored:", index.ntotal)
````

````python
# ================================
# Save FAISS index
# ================================
faiss.write_index(index, FAISS_PATH)
print("FAISS index saved to:", FAISS_PATH)
````

````python
def search(query, top_k=5):
    """
    Perform semantic search over FAISS index for a given query.

    Args:
        query (str): The user query.
        top_k (int): Number of top results to return.

    Returns:
        List[dict]: Top-k chunks with content, metadata, and score.
    """
    if not query:
        return []

    # 1. Encode query
    query_embedding = model.encode(
        ["passage: " + query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # 2. Search FAISS index
    scores, indices = index.search(query_embedding, top_k)

    # 3. Collect results
    results = []
    for idx, score in zip(indices[0], scores[0]):
        if idx < 0 or idx >= len(chunk_objects):
            continue  # Skip invalid indices

        chunk = chunk_objects[idx]
        results.append({
            "content": chunk.get("content") or chunk.get("text"),
            "metadata": chunk.get("metadata", {}),
            "source": chunk.get("source", None),
            "chunk_id": chunk.get("chunk_id", idx),
            "score": float(score)
        })

    return results
````
